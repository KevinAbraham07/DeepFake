## 2. Link your Dataset
1. On the right side panel of Kaggle, click **Add Data**.
2. Search for `140k Real and Fake Faces` and add it.
3. Open `train.py` (Line 61) and update the `dataset_path` to point to the Kaggle input folder (e.g., `/kaggle/input/real-and-fake-face-detection/real_and_fake_face/`).
